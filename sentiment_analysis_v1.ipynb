```python
# reference: https://textblob.readthedocs.io/en/dev/quickstart.html
# pip install nltk textblob on jupyter
# once install, to download data use this cmd: python -m textblob.download_corpora on anaconda prompt
import logging
from textblob import TextBlob
import nltk
nltk.download('punkt_tab')
nltk.download('punkt')
```

```python
# Configure logging
logging.basicConfig(level=logging.INFO,
                    format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)
```

```python
# Create a TextBlob object
text = 'US attacks Venezuela, analyst says crude prices may go down'
blob = TextBlob(text)
```

```python
print(blob.sentiment)
```

```python
# Tokenization
logger.info(blob.words)
logger.info(blob.sentences)
```

```python
# Part-of-Speech Tagging
logger.info(blob.tags)
```

```python
# Noun Phrases
logger.info(blob.noun_phrases)
```

```python
# Sentiment Analysis
print(blob.sentiment)
```

```python
def analyze_sentiment(text):
    # Create a TextBlob object
    blob = TextBlob(text)
    # Determine the sentiment polarity
    sentiment_polarity = blob.sentiment.polarity
    # Determine the sentiment subjectivity
    sentiment_subjectivity = blob.sentiment.subjectivity
    # Output the results
    return {'polarity': sentiment_polarity, 'subjectivity': sentiment_subjectivity}
```

```python
# Example usage
if __name__ == '__main__':
    text = 'I love programming with Python. Its so much fun and versatile!'
    # text = 'Warren Buffett fears what the market doesnt: tax hikes. And he has little room to maneuver'
    # text = 'China Vows to Accelerate Buying Unsold Homes for Public Housing'
    # text = 'Trump and India Call Off Their Trade War, but the Terms of Peace Are Murky'
    # subjectivity: low score indicates its more of a objective or factual
    sentiment = analyze_sentiment(text)
    logger.info(f'Text: {text}')
    logger.info(f'Sentiment Polarity: {sentiment["polarity"]}')
    logger.info(f'Sentiment Subjectivity: {sentiment["subjectivity"]}')
```
